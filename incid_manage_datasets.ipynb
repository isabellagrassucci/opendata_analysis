```python
def data_months_check (path, directory):
    base_dir = path
    inc_dir = os.path.join(path, directory)
    print(inc_dir)

    csv_files = [f for f in os.listdir(inc_dir) if f.endswith(".csv")]

    print(csv_files)

    list_columns = list(("Protocollo", "Gruppo", "DataOraIncidente", "Localizzazione1", "STRADA1", 'particolaritastrade',
                         'TipoStrada', 'FondoStradale', 'Pavimentazione', 'Segnaletica', 'Illuminazione',
                         'Longitude', 'Latitude', 'TipoVeicolo', 'Sesso', 'CinturaCascoUtilizzato', 'Airbag'))
    
    for file in csv_files:
        print("\n"f"FILE: {file}")
        
        file_path = os.path.join(inc_dir, file)
        df = pd.read_csv(file_path, encoding="ISO-8859-1", header=0, delimiter=";")#, usecols = list_columns)
        df.drop(df.columns[df.columns.str.contains('unnamed', case=False)], axis=1, inplace=True)


        df["DataOraIncidente"] = pd.to_datetime(
            df["DataOraIncidente"],
            format="%d/%m/%Y %H:%M:%S",
            errors="coerce"
        )
        df["DataIncidente"] = df["DataOraIncidente"].dt.date
        df["OraIncidente"] = df["DataOraIncidente"].dt.time
    
        #print (df["DataIncidente"].dropna().unique())
    
        mesi_presenti = ( df["DataOraIncidente"] .dropna() .dt.to_period("M") .astype(str) .unique() )
        
        print("\nMesi presenti (YYYY-MM):")
        print(sorted(mesi_presenti))

        #And rename files based on the months


        print("\nColonne:")
        print(df.columns.tolist())
        
        
        # Prendiamo la prima data valida
        first_date = df["DataOraIncidente"].dropna().iloc[0]
        
        year = first_date.year
        month = f"{first_date.month:02d}"  # zero-padding
        
        new_name = f"incidenti_{year}_{month}.csv"
        new_path = os.path.join(inc_dir, new_name)
        
        print(f"{file}  →  {new_name}")
        
        #os.rename(path, new_path)
        
        
#print(df.head())
```

```python
base_dir = "/mnt/d/isabe/Desktop/Progetti_ripresa_lavoro/data/2022/"
inc_dir = "incidenti/csv"
data_months_check(base_dir, inc_dir)
```

```python
#Last function also renames the files. This function would take the renamed files and merge them.
def create_file_by_year (path, directory):
    base_dir = path
    inc_dir = os.path.join(path, directory)
    print(inc_dir)
    
    all_files = sorted([
        f for f in os.listdir(inc_dir)
        if f.startswith("incidenti_2022_") and f.endswith(".csv")
    ])
    print(all_files)

    dfs = []

    #Use all the columns that has at least a menaning. The distribution will be check in the feature selection
    list_columns = list(("Protocollo", "Gruppo", "DataOraIncidente", 'particolaritastrade',
                         'TipoStrada', 'FondoStradale', 'Pavimentazione', 'Segnaletica',
                         'CondizioneAtmosferica', 'Traffico', 'Visibilita', 'Illuminazione',
                         'NUM_FERITI', 'NUM_RISERVATA', 'NUM_MORTI', 'NUM_ILLESI', 'Longitude',
                         'Latitude', 'TipoVeicolo', 'Sesso', 'Deceduto', 'DecedutoDopo', 'CinturaCascoUtilizzato', 'Airbag'))
    
    for file in all_files:
        path = os.path.join(inc_dir, file)
        
        df = pd.read_csv(
            path,
            encoding="ISO-8859-1",
            delimiter=";",
            usecols = list_columns
        )
        #df.drop(df.columns[df.columns.str.contains('unnamed', case=False)], axis=1, inplace=True)
        
        dfs.append(df)
    
    incidenti_2022 = pd.concat(dfs, ignore_index=True)
    
    output_path = os.path.join(inc_dir, "incidenti_2022.csv")
    incidenti_2022.to_csv(output_path, index=False, encoding="ISO-8859-1")
    
    print("Shape finale:", incidenti_2022.shape)
    print(incidenti_2022.columns)
    #display(incidenti_2022.iloc[0:5, 9:-1])
    display(incidenti_2022.head())
    display(incidenti_2022.tail())

```

```python
base_dir = "/mnt/d/isabe/Desktop/Progetti_ripresa_lavoro/data/2022/"
inc_dir = "incidenti/csv"
create_file_by_year(base_dir, inc_dir)
```
